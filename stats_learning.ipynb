```python
import scipy.stats as stats

old_layout_spend = [45, 52, 38, 41, 60, 49, 44, 55, 48, 50]
new_layout_spend = [58, 62, 51, 55, 67, 60, 53, 59, 64, 61]


t_stat, p_value = stats.ttest_ind(old_layout_spend, new_layout_spend)

print(f"P-value: {p_value:.5f}")

if p_value < 0.05:
  print("Result: The new layout significantly increases spending! (Reject Null Hypothesis)")
else:
  print("Result: No significant difference between the two layouts.")
```

```python
import scipy.stats as stats

sample_rods = [49.8, 50.1, 49.9, 49.7, 50.2, 50.0, 49.8, 49.6, 49.9]

target_length = 50.0

t_stat, p_value = stats.ttest_1samp(sample_rods, target_length)

print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
  print("Result: Machine is misaligned! Rod lengths are significantly off target.")
else:
  print("Result: Machine is working fine. Variations are just random noise.")
```

```python
import numpy as np
import scipy.stats as stats

coffee_data = np.array([
    [120, 30], 
    [80, 100], 
])

chi2, p_value, dof, expected = stats.chi2_contingency(coffee_data)

print(f"P-value: {p_value:.5f}")

if p_value < 0.05:
  print("Result: Membership status and pastry purchases are dependent (There is a relationship).")
else:
  print("Result: Membership status and pastry purchases are independent (No relationship).")
```

```python
import numpy as np
from statsmodels.stats.proportion import proportion_confint, proportions_ztest

conversions = np.array([50, 70])
nobs = np.array([1000, 1000])

# Calculate conversion rates
conv_rates = conversions / nobs
print(f"Control Conversion Rate: {conv_rates[0]*100:.2f}%")
print(f"Variant Conversion Rate: {conv_rates[1]*100:.2f}%")
print(f"Absolute Lift: {(conv_rates[1] - conv_rates[0])*100:+.2f}%\n")

# 2. Perform a Two-Proportion Z-Test (to get the P-value)
z_stat, p_value = proportions_ztest(conversions, nobs)

print(f"Z-statistic: {z_stat:.4f}")
print(f"P-value: {p_value:.4f}")

ci_control = proportion_confint(
    conversions[0], nobs[0], alpha=0.05, method='normal'
)
ci_variant = proportion_confint(
    conversions[1], nobs[1], alpha=0.05, method='normal'
)

print(
    f"Control 95% CI: ({ci_control[0]*100:.2f}%, {ci_control[1]*100:.2f}%)"
)
print(
    f"Variant 95% CI: ({ci_variant[0]*100:.2f}%, {ci_variant[1]*100:.2f}%)"
)

# 4. Final Decision Rule
alpha = 0.05
if p_value < alpha:
  print(
      "\nResult: 🎉 Statistically Significant! Roll out the new variant."
      f" (P-value {p_value:.4f} < {alpha})"
  )
else:
  print(
      "\nResult: ❌ Not Significant. Keep the old version or run the test"
      f" longer. (P-value {p_value:.4f} >= {alpha})"
  )
```
